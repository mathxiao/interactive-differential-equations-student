# Student Guide — Interactive Differential Equations Notebooks

These notebooks are designed for use during class and for independent review.

## During class

- Open the lecture notebook for the current session.
- Run cells from top to bottom.
- When you reach a **Classroom Checkpoint**, stop and work the question before moving on.
- Interactive sliders and plots are intended for experimentation: change parameters and observe what changes mathematically.

## Homework

Each lecture has a matching homework notebook. Some computational exercises contain `TODO` cells for you to complete.

## Answers and solutions

This student repository intentionally does **not** include instructor answer-reveal cells or homework solution notebooks.

## Cursor, JupyterLab, and VS Code

The `.ipynb` files can be opened in any of these environments. Use the Python environment created from `requirements.txt`.
