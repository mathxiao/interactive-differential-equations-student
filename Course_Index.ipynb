# Differential Equations — Student Interactive Course
## Chapters 1–9

This repository contains the student-facing lecture and homework notebooks for the course.

### How to use it

1. Open the notebook for the current class session.
2. Run the local setup cell first.
3. Read and execute cells sequentially.
4. Stop at each **Classroom Checkpoint** and work the question before discussion.
5. Complete the matching homework notebook after class.

**No instructor solutions or instructor answer-reveal cells are included in this package.**
